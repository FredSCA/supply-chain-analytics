# Generación Automática de Reportes con IA Generativa

## 📌 Objetivo de Negocio
Transformar los resultados cuantitativos del análisis de datos (Módulo 1) en un **Resumen Ejecutivo en lenguaje natural** utilizando Modelos de Lenguaje Grande (LLMs). Esto permite a los gerentes y tomadores de decisiones comprender rápidamente el estado de la operación sin necesidad de leer código o tablas complejas.

```python
# Instalar la nueva librería oficial de Google
!pip install -q google-genai python-dotenv
```

```python
from google import genai
import os
from dotenv import load_dotenv

# Cargar variables de entorno
load_dotenv()

# Configurar el cliente
api_key = os.getenv("GOOGLE_API_KEY")
client = genai.Client(api_key=api_key)

print("✅ API Key configurada con la nueva librería google-genai.")
```

## 1. Preparación del Contexto (Prompt Engineering)
Para que la IA genere un reporte útil, le proporcionamos los datos exactos extraídos en el análisis exploratorio previo.

```python
# Contexto de datos extraído del Módulo 1
contexto_datos = """
Datos Operativos del Almacén de Tránsito (Dinet - Mina):
- El tiempo de tránsito promedio es de 1 día (24 horas).
- Se detectaron y eliminaron 1,687 registros con errores de digitación en fechas.
- Los meses con mayor volumen de despacho son Septiembre, Octubre y Noviembre (picos de hasta 1,014 viajes).
- Los días de la semana con mayor carga operativa son los Lunes y Martes.
- El objetivo del Módulo 2 fue predecir si un envío será de gran o pequeño volumen, logrando una precisión del 77%.
"""

# Diseño del Prompt (Instrucción para la IA)
prompt = f"""
Actúa como un Analista Senior de Supply Chain. A continuación te presento los hallazgos de un análisis de datos de nuestro almacén de tránsito:

{contexto_datos}

Por favor, redacta un "Resumen Ejecutivo" formal, directo y orientado a la acción (máximo 200 palabras). 
El reporte debe incluir:
1. Una introducción breve sobre la eficiencia del almacén.
2. Los 3 hallazgos más importantes (menciona la estacionalidad y la calidad de los datos).
3. Dos recomendaciones estratégicas para la gerencia basadas en esta información.
"""

print("✅ Prompt preparado con éxito.")
```

## 2. Llamada a la IA Generativa
Enviamos el contexto al modelo y obtenemos el reporte redactado automáticamente.

```python
# Generar la respuesta usando el nuevo cliente
response = client.models.generate_content(
    model='gemini-3.5-flash-lite',  # Modelo actualizado para usuarios nuevos
    contents=prompt,
)

print("==================================================")
print("🤖 REPORTE GENERADO POR IA:")
print("==================================================")
print(response.text)
```
